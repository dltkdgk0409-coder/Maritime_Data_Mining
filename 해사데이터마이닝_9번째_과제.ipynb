```python
# 문제 5 - 카이제곱 독립성 검정: 유입채널과 구매여부의 관련성
from pathlib import Path
import pandas as pd
from scipy.stats import chi2_contingency

DATA_DIR = Path('/content')
df = pd.read_csv(DATA_DIR / 'channel_purchase.csv')

# 1. 데이터 확인
print('[데이터 상위 5행]')
print(df.head(), '\n')
print('[변수 정보]')
print(df.info(), '\n')

# 2. 교차표 생성
table = pd.crosstab(df['channel'], df['purchase_yn'])
print('[교차표]')
print(table, '\n')

# 3. 카이제곱 독립성 검정
chi2, p, dof, expected = chi2_contingency(table)
print('[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

# 4. 기대도수 확인
expected_df = pd.DataFrame(expected, index=table.index, columns=table.columns)
print('\n[기대도수]')
print(expected_df)

# 5. 채널별 구매전환율
conversion_rate = df.groupby('channel')['purchase_yn'].mean().sort_values(ascending=False) * 100
print('\n[채널별 구매전환율(%)]')
print(conversion_rate.round(2))

# 6. 해석
print('\n[해석]')
if p < 0.05:
    print('p-value가 0.05보다 작으므로, 유입채널과 구매여부는 독립이 아니며 서로 관련이 있다고 해석할 수 있다.')
else:
    print('p-value가 0.05 이상이므로, 유입채널과 구매여부의 관련성을 확인하기 어렵다.')
print(f'가장 높은 구매전환율 채널: {conversion_rate.index[0]} ({conversion_rate.iloc[0]:.2f}%)')
print(f'가장 낮은 구매전환율 채널: {conversion_rate.index[-1]} ({conversion_rate.iloc[-1]:.2f}%)')
```

```python
# 문제 6 - 카이제곱 동질성 검정: 성별에 따른 콘텐츠 선호 분포 비교
from pathlib import Path
import pandas as pd
from scipy.stats import chi2_contingency

DATA_DIR = Path('/content')
df = pd.read_csv(DATA_DIR / 'gender_preference.csv')

# 1. 데이터 확인
print('[데이터 상위 5행]')
print(df.head(), '\n')
print('[변수 정보]')
print(df.info(), '\n')

# 2. 교차표 생성
table = pd.crosstab(df['gender'], df['content_type'])
print('[교차표]')
print(table, '\n')

# 3. 카이제곱 동질성 검정
chi2, p, dof, expected = chi2_contingency(table)
print('[검정 결과]')
print('chi-square statistic:', round(chi2, 4))
print('p-value:', round(p, 6))
print('degrees of freedom:', dof)

# 4. 기대도수 확인
expected_df = pd.DataFrame(expected, index=table.index, columns=table.columns)
print('\n[기대도수]')
print(expected_df)

# 5. 성별별 최다 선호 유형
top_pref = table.idxmax(axis=1)
print('\n[성별별 최다 선호 콘텐츠]')
print(top_pref)

# 6. 성별 내 비율
row_ratio = pd.crosstab(df['gender'], df['content_type'], normalize='index') * 100
print('\n[성별 내 콘텐츠 선호 비율(%)]')
print(row_ratio.round(2))

# 7. 해석
print('\n[해석]')
if p < 0.05:
    print('p-value가 0.05보다 작으므로, 성별에 따라 콘텐츠 선호 분포가 다르다고 해석할 수 있다.')
else:
    print('p-value가 0.05 이상이므로, 성별에 따른 콘텐츠 선호 분포 차이를 확인하기 어렵다.')
for g in top_pref.index:
    print(f'{g}의 최다 선호 콘텐츠: {top_pref[g]}')
```
