```python
# 문제 3 - 일원분산분석: 지점별 고객만족도 비교
import pandas as pd
from scipy import stats

df = pd.read_csv("branch_satisfaction.csv")

# Q3. 지점별 데이터 분리 및 일원 ANOVA 수행
a = df[df["branch"] == "A"]["score"]
b = df[df["branch"] == "B"]["score"]
c = df[df["branch"] == "C"]["score"]

result = stats.f_oneway(a, b, c)

# Q4. 지점별 평균 계산
means = df.groupby("branch")["score"].mean()
print("지점별 평균:")
print(means)
print("F-statistic:", result.statistic)
print("p-value:", result.pvalue)
print("평균이 가장 높은 지점:", means.idxmax())

# Q5. 결론
if result.pvalue < 0.05:
    print("결론: p-value < 0.05이므로 지점 간 평균 고객만족도는 통계적으로 유의한 차이가 있다.")
    print("후속 조치: 만족도가 낮은 지점을 대상으로 서비스 품질 개선 프로그램을 도입할 필요가 있다.")
else:
    print("결론: p-value >= 0.05이므로 지점 간 평균 만족도 차이가 있다고 보기 어렵다.")
```

```python
# 문제 4 - 이원분산분석: 광고유형과 연령대가 구매금액에 미치는 영향
import pandas as pd
import statsmodels.formula.api as smf
from statsmodels.stats.anova import anova_lm

df = pd.read_csv("ad_age_sales.csv")

# Q2. 이원 ANOVA 수행 (주효과 + 상호작용 포함)
model = smf.ols("sales_amount ~ C(ad_type) + C(age_group) + C(ad_type):C(age_group)", data=df).fit()
anova_table = anova_lm(model, typ=2)
print(anova_table)

# Q3. 광고유형 주효과 판단
ad_pvalue = anova_table.loc["C(ad_type)", "PR(>F)"]
print("\n광고유형 p-value:", ad_pvalue)
if ad_pvalue < 0.05:
    print("→ 광고유형에 따라 평균 구매금액 차이가 유의하다.")
else:
    print("→ 광고유형에 따른 구매금액 차이가 유의하지 않다.")

# Q4. 연령대 주효과 판단
age_pvalue = anova_table.loc["C(age_group)", "PR(>F)"]
print("\n연령대 p-value:", age_pvalue)
if age_pvalue < 0.05:
    print("→ 연령대에 따라 평균 구매금액 차이가 유의하다.")
else:
    print("→ 연령대에 따른 구매금액 차이가 유의하지 않다.")

# Q5. 상호작용 효과 판단
inter_pvalue = anova_table.loc["C(ad_type):C(age_group)", "PR(>F)"]
print("\n상호작용 p-value:", inter_pvalue)
if inter_pvalue < 0.05:
    print("→ 광고유형의 효과는 연령대에 따라 달라지므로, 연령대별 맞춤형 광고 전략이 필요하다.")
else:
    print("→ 광고유형과 연령대 간 상호작용 효과는 유의하지 않다.")
```
