```python
# 문제 1 - 단일표본 t-test: 배달시간 평균 검정
import pandas as pd
from scipy import stats
import numpy as np

df = pd.read_csv("delivery_time.csv")
x = df["delivery_min"].dropna()

# Q3. 단일표본 t-test (대립가설: 평균 < 30)
result = stats.ttest_1samp(x, popmean=30, alternative="less")

# Q4. 표본평균과 95% 신뢰구간
mean_x = x.mean()
n = len(x)
se = stats.sem(x)
ci_low, ci_high = stats.t.interval(0.95, df=n-1, loc=mean_x, scale=se)

print("표본평균:", mean_x)
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)
print("95% CI:", (ci_low, ci_high))

# Q5. 결론
if result.pvalue < 0.05 and mean_x < 30:
    print("결론: p-value < 0.05이고 표본평균이 30분보다 작으므로, 평균 배달시간은 30분 미만이라고 통계적으로 해석할 수 있다.")
else:
    print("결론: 평균 배달시간이 30분 미만이라고 통계적으로 결론내리기 어렵다.")
```

```python
# 문제 2 - 대응표본 t-test: 교육 전후 업무처리시간 비교
import pandas as pd
from scipy import stats

df = pd.read_csv("before_after_worktime.csv")
before = df["before_min"]
after = df["after_min"]

# Q4. 평균 차이 계산
diff = before - after
print("교육 전 평균:", before.mean())
print("교육 후 평균:", after.mean())
print("평균 차이(before - after):", diff.mean())

# Q3. 대응표본 t-test (대립가설: before > after, 즉 교육 후 감소)
result = stats.ttest_rel(before, after, alternative="greater")
print("t-statistic:", result.statistic)
print("p-value:", result.pvalue)

# Q5. 결론
if result.pvalue < 0.05 and diff.mean() > 0:
    print("결론: p-value < 0.05이고 평균 차이가 양수이므로, 교육 후 업무처리시간이 유의하게 감소했다고 볼 수 있다.")
else:
    print("결론: 교육 후 업무처리시간이 감소했다고 통계적으로 말하기 어렵다.")
```
